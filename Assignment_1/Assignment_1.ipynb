```python
import tensorflow as tf
import  matplotlib.pyplot as plt
import numpy as np
```

```python
fashion_mnist = tf.keras.datasets.fashion_mnist

(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
```

```python
print("tarining shape",train_images.shape)
print("tarining labels",train_labels.shape)
print("testing shape",test_images.shape)
print("testing labels",test_labels.shape)
```

```python
train_images = train_images.astype("float32")
test_images = test_images.astype("float32")
```

```python
train_images = train_images / 255.0
test_images = test_images / 255.0
```

```python
print(train_images.min())
print(train_images.max())
```

```python
print("Training Samples:", len(train_images))
print("Testing Samples:", len(test_images))
```

```python
class_names = [
    'T-shirt/top',
    'Trouser',
    'Pullover',
    'Dress',
    'Coat',
    'Sandal',
    'Shirt',
    'Sneaker',
    'Bag',
    'Ankle boot'
]
```

```python
plt.imshow(train_images[0], cmap='gray')
plt.title(class_names[train_labels[0]])
plt.colorbar()
plt.show()
```

```python
plt.figure(figsize=(8,8))

for i in range(16):
    plt.subplot(4,4,i+1)
    plt.imshow(train_images[i], cmap='gray')
    plt.title(class_names[train_labels[i]])
    plt.axis("off")

plt.show()
```

```python
unique, counts = np.unique(train_labels, return_counts=True)

plt.bar(unique, counts)
plt.xticks(unique, class_names, rotation=90)
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.title("Class Distribution")
plt.show()
```
